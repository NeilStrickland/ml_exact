```python
examples = [
    (PartialSumModel(), [2 ** i for i in range(5)]),
    (ResettingPartialSumModel(), [1, 10, 100, 0, 1, 2, 4, -1, -2, 11111, 1110, 100]),
]
for model, values in examples:
    print(values)
    x = ops.reshape(ops.convert_to_tensor(values, dtype='float32'), (1, -1, 1))
    y = model(x)
    result = ops.convert_to_numpy(y)
    assert isinstance(result, np.ndarray)
    print(result.ravel().tolist())
    print()
```
